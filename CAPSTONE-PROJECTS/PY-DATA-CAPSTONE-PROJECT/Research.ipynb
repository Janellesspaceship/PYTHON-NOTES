# Research on the Kenya Finance Bill 2024
## Introduction

The Kenya Finance Bill 2024 was one of the most controversial legislative proposals in Kenya’s recent history. Introduced in May 2024 by the government of William Ruto, the bill aimed to increase government revenue through new taxes and amendments to existing tax laws. The government argued that the bill was necessary to reduce Kenya’s growing debt burden and fund development projects. However, many Kenyans viewed it as economically harsh during a period of high living costs, unemployment, and inflation.

The bill triggered nationwide protests led largely by Gen Z and young Kenyans under the hashtag #RejectFinanceBill2024. Demonstrations escalated into one of the biggest youth-led movements in Kenyan history, eventually forcing President Ruto to withdraw the bill.

## Background

Kenya has been struggling with:

- Rising public debt
- High taxation
- Inflation and increased cost of living
- Pressure from international lenders such as the IMF

The government sought to raise approximately KSh 346 billion through the Finance Bill 2024 in order to reduce the fiscal deficit and meet debt obligations.

The bill was also linked to economic reform conditions supported by the International Monetary Fund and the World Bank. Critics argued that ordinary citizens were being forced to carry the burden of government debt and overspending.

**Reasons for Increase in Fuel Prices in Kenya (2024–2026)**

Fuel prices in Kenya have risen significantly since 2024 due to a combination of global shocks, currency depreciation, and domestic cost structures.

`1. Rising global crude oil prices`

One of the main drivers is the increase in international crude oil prices.

- Kenya imports almost all its fuel.

- Global supply disruptions (especially Middle East conflicts and shipping route instability) have pushed crude oil prices higher.

- When crude oil prices rise, the landing cost of fuel in Kenya increases directly.

This is why EPRA often cites “international market prices” during fuel hikes.

`2. Geopolitical conflicts affecting supply chains`

Since 2024–2026:

- Conflicts in the Middle East (major oil-producing region) have disrupted supply routes.

- Shipping delays and tanker rerouting (e.g., Red Sea instability) increased transport costs.

- Panic buying during crisis periods has also contributed to short-term spikes.

These disruptions reduced supply stability and increased global fuel volatility.

`3. Weakening of the Kenyan Shilling`

The exchange rate plays a major role because:

- Fuel is bought in US dollars

- A weaker Kenyan shilling means Kenya pays more in local currency

Even if global prices remain stable, a weaker shilling still makes fuel more expensive locally.

`4. Taxes and levies on fuel`

Fuel prices in Kenya are not only about crude oil. They include:

- VAT

- Excise duty

- Road maintenance levies

- Distribution and marketing margins

These taxes make up a significant portion of pump prices, so any adjustment increases retail fuel costs.

`5. Higher import and logistics costs`

Fuel passes through:

- Shipping (freight costs increased globally)

- Port handling at Mombasa

- Storage and distribution

When global shipping costs rise, Kenya’s landed cost increases, pushing pump prices up.

`6. Government pricing formula (EPRA monthly review)`

Kenya uses a monthly fuel pricing mechanism where EPRA adjusts prices based on:

- global crude prices

- exchange rate changes

- landing and transport costs

- taxes and margins

This means fuel prices respond quickly to global changes rather than being fixed.

`7. Inflation feedback loop`

Fuel price increases also become self-reinforcing:

- Higher fuel → higher transport costs

- Higher transport → higher food and goods prices

- This increases overall inflation, which further pressures the economy
