#### Creating Arrays with the `np.zeros()` function

- `np.zeros` is a NumPy function used to create a new array filled entirely with zeros.
- Useful when:
    - You need placeholder values
    - You're preparing storage for data
    - You're building matrices/grids
    - You want a default starting point before calculations  

- Default dtype is `float`    

```python
# Creating a 1D Array
np.zeros(5)
```

```python
# Creating a 2D Array (Rows and Columns)
np.zeros((6, 3))
```

```python
# Creating Integer Zeros
# Have an array with zeros but as integer values
np.zeros((8, 4), dtype = int)
```

```python
# Creating a 3D Array
np.zeros((2, 3, 4))
```

#### Creating Arrays with the `np.arange()` function

- `np.arange()` is a NumPy function used to create arrays with a sequence of evenly spaced values over a specified range.
- Useful when:
    - Number sequences
    - Counting patterns
    - Even/odd numbers
    - Time intervals
    - Index values
    - Mathematical datasets

- SYNTAX: `np.arange(start, stop, step)`
   - `start` - where the sequence begins
   - `stop` - where the sequence ends`(not included)`    
   - `step` - how much each number increases`(default=1)`

```python
np.arange(5)

# starts at 0 by default
# stops before 5
# step is 1
```

`Start and Stop`

```python
np.arange(2, 10)
```

`Using Step Values`

```python
np.arange(0, 11, 2)

# Starts at 0
# Stops before 11
# Increases by 2
# Great for even numbers
```

`Odd Numbers`

```python
np.arange(1, 11, 2)

```

`Negative Step(Counting Backwards)`

```python
np.arange(10, 0, -2)

# Starts at 10
# Decreases by 2
# Useful for countdowns
```

`Decimal Steps`

```python
np.arange(0, 1, 0.2)

# Used in plotting graphs
# Used in scientific calculations
# Used for probability scales
```

```python
# Array with 50's

np.arange(50, 60, 2)
```

```python
plt.scatter(np.arange(0, 7),
            np.arange(-3, 4))
plt.show()
```

#### Shape Shifting

- We have seen that we can use the shape attribute to get the shape of an array 

- What if we want to change the shape of the array 

- We know that in 2D arrays 
    - Rows are the first dimension 
    - Coulumns are the second dimension

#### Flattening an Array

- Removing an array from whichever dimension it is into a 1D Array

```python
sample_array = np.array([[12, 34, 50],
                         [23, 45, 67],
                         [34, 56, 78]])

sample_array.shape
```

```python
flattened_array = sample_array.flatten()
flattened_array.shape
```

#### Reshaping an Array

- Changing the shape of an array to a desired shape

**Rules for Reshaping an Array:**
- Rule 1 : The sum of values has to be the same as the product of the shape values you want to change into. 
- Rule 2

```python
counts_of_fireworks = np.array([[1, 2],
                                [3, 4],
                                [5, 6]])
counts_of_fireworks.shape
```

```python
new_array = counts_of_fireworks.reshape(2, 3)
new_array
```

## Fancy Indexing

- This is a Numpy technique that allows you to select `multiple specific elements`, `rows`, or `columns` using `lists` or `arrays of indices` instead of slices.

#### `Normal Indexing:`

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr[3])
```

#### `Fancy Indexing:`

```python
print(arr[[0, 1, 4]])
```

#### `Fancy Indexing with Repeated Values:`

```python
print(arr[[2, 2, 4]])     # Index 2 was selected twice
```

#### `Reordering Elements:`

```python
print(arr[[4, 3, 2, 1, 0]])      # You can rearrange data
```

### `Fancy Indexing in 2D Arrays:`

```python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(matrix)
```

#### `1.Selecting Specific Rows`

```python
print(matrix[[0, 2]])
```

#### `2.Selecting Specific Columns`

```python
# To select columns, use `:` for rows and index list for columns
# **:** - means all rows

print(matrix[:, [0, 2]])
```

#### `3.Selecting Specific Individual Elements`

```python
print(matrix[[0, 1, 2], [2, 1, 0]])

# It pairs indices
# 30 - (0, 2)
# 50 - (1, 1)
# 70 - (2, 0)
```

#### `Boolean Fancy Indexing(Filtering)`

```python
arr = np.array([10, 15, 20, 25, 30, 35, 40, 45, 50])
print(arr[arr>20])    # Only values greater than 20 are selected
```

#### `Multiple Conditions`

`&` - AND

`|` - OR

```python
print(arr[(arr>20) & (arr<50)])
```

```python
print(arr[(arr>20) | (arr<50)])
```
